# 9. Handling Missing Values

```python
df[["Age","Annual_Salary","Experience_Years","Performance_Score"]].isnull().sum()
```

```python
# Fill NAN Values of Age with median computation
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Age"].isnull().sum()
```

```python
# For Annual Salary
df["Annual_Salary"] = df["Annual_Salary"].fillna(df["Annual_Salary"].median())
```

```python
df["Annual_Salary"].isnull().sum()
```

```python
# Experience Years
df["Experience_Years"] = df["Experience_Years"].fillna(df["Experience_Years"].median())
```

```python
df["Experience_Years"].isnull().sum()
```

```python
# Performance Score
df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].median()
)
```

```python
df["Performance_Score"].isnull().sum()
```

# 10. Handling Missing Categorical Values

```python
df[["Department","Gender","City","Work_Mode"]].isnull().sum()
```

```python
# Department
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)
```

```python
# Gender
df["Gender"] = df["Gender"].fillna(
    df["Gender"].mode()[0]
)
```

```python
# City
df["City"] = df["City"].fillna(
    df["City"].mode()[0]
)
```

```python
# Work Mode
df["Work_Mode"] = df["Work_Mode"].fillna(
    df["Work_Mode"].mode()[0]
)
```

```python
df[["Department","Gender","City","Work_Mode"]].isnull().sum()
```
